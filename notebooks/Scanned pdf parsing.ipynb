```python
import pdfplumber

MIN_TEXT_LENGTH = 100

def extract_from_unscanned():
    pdf_path = r"/home/san/Documents/Important/cv.pdf"
    text = ""
    with pdfplumber.open(pdf_path) as pdf:
        for page in pdf.pages:
            page_text = page.extract_text()
            if page_text:
                text += page_text + "\n"
            
        print(text)
            
```

```python
from pdf2image import convert_from_path
import easyocr
import numpy as np
import cv2

def extract_from_scanned():
    pdf_path = r"/home/san/Documents/Important/SanCV.pdf"
    images = convert_from_path(pdf_path)
    text = ""
    reader = easyocr.Reader(['en'])
    
    for image in images:
        image_np = np.array(image)
        inverted_image = cv2.bitwise_not(image_np)
        result = reader.readtext(image_np,detail = 0)
        text += " ".join(result)+'\n'
    return text.strip()
```
